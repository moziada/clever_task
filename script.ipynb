```python
def get_regions_mapping(
        regions: pd.Series,
        regions_key: pd.DataFrame,
) -> pd.Series:
    key_row_mapper = regions_key[['key_row', 'zillow_region_name']]
    key_row_mapper.drop_duplicates(['key_row'], keep='first', inplace=True)
    key_row_mapper = key_row_mapper.set_index('key_row')
    key_row_mapper = key_row_mapper.to_dict()['zillow_region_name']
    # mapped_state_names = output_df['key_row'].map(key_row_mapper)
    mapped_regions = regions.map(key_row_mapper)
    return mapped_regions

def get_census_population(
        census_population: pd.DataFrame,
        mapped_regions: pd.Series
) -> np.ndarray:
    # Getting all regions population data
    full_population_series = census_population[census_population.iloc[:, 0] == "    Total population"].iloc[0]
    
    region_col_names = mapped_regions + "!!Estimate"
    regions_population = full_population_series[region_col_names].values
    return regions_population

def get_median_household_income(
        census_mhi: pd.DataFrame,
        mapped_regions: pd.Series
) -> np.ndarray:
    full_series = census_mhi[census_mhi.iloc[:, 0] == "    Households"].iloc[0]
    
    region_col_names = mapped_regions + "!!Median income (dollars)!!Estimate"
    return full_series[region_col_names].values

# Function to convert number to ordinal
def ordinal(n):
    if 10 <= n % 100 <= 20:
        suffix = 'th'
    else:
        suffix = {1: 'st', 2: 'nd', 3: 'rd'}.get(n % 10, 'th')
    return f"{n}{suffix}"

def get_rank(
        numbers_series: pd.Series
) -> pd.Series:
    numbers_series = numbers_series.copy()
    # clean str column and convert to number datatype if it is not numeric
    if not pd.api.types.is_numeric_dtype(numbers_series):
        numbers_series = pd.to_numeric(numbers_series.str.replace(r'[^0-9.]', ''))   # clean symbols like `$`, `,` and convert to numeric 
    rankings = numbers_series.rank(ascending=False, method='min').astype(int)
    rankings_with_suffix = rankings.apply(ordinal)
    return rankings_with_suffix

def generate_blurb(
        df: pd.DataFrame,
        blurb_template: str,
        **columns
):
    """
    Generates a blurb using any number of columns mapped to template placeholders.

    Parameters:
    - df: pandas DataFrame
    - template: string with named placeholders (e.g. "{city} is {rank}")
    - **columns: mapping of template placeholders to DataFrame column names

    Returns:
    - pandas Series of generated blurbs
    """
    return df.apply(lambda row: blurb_template.format(**{k: row[v] for k, v in columns.items()}), axis=1)

output_df = pd.DataFrame()
output_df['key_row'] = KEY_ROWS
output_df['mapped_key_row'] = get_regions_mapping(regions=output_df['key_row'], regions_key=regions_key_df)

output_df['census_population'] = get_census_population(census_population=census_population_df, mapped_regions=output_df['mapped_key_row'])
output_df['population_rank'] = get_rank(output_df['census_population'])
output_df['population_blurb'] = generate_blurb(
    df=output_df,
    blurb_template="{region} is {rank} in the nation in population among states, DC, and Puerto Rico.",
    region="mapped_key_row", rank='population_rank'
)

output_df['median_household_income'] = get_median_household_income(census_mhi=census_mhi_df, mapped_regions=output_df['mapped_key_row'])
output_df['median_household_income_rank'] = get_rank(output_df['median_household_income'])
output_df['median_household_income_blurb'] = generate_blurb(
    df=output_df,
    blurb_template="{region} is {rank} in the nation in median household income among states, DC, and Puerto Rico.",
    region="mapped_key_row", rank='population_rank'
)

currency_cols = ['median_household_income', 'median_sale_price']
output_df
```
