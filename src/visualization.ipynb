```python
import pandas as pd
import plotly.express as px

df = pd.read_excel('../output/OUTPUT.xlsx')
df.head()
```

```python
df['census_population'] = pd.to_numeric(df['census_population'])
df['median_household_income'] = pd.to_numeric(df['median_household_income'])
df['median_sale_price'] = pd.to_numeric(df['median_sale_price'])
```

```python
fig = px.bar(
    df,
    x='key_row',
    y='census_population',
    title='Population by Region',
    labels={'key_row': 'Region', 'census_population': 'Population'},
)
fig.update_layout(xaxis_tickangle=-65)
fig.show()
```

The plot highlights states like California, Florida, New York, and Texas are among the high-density regions. If these regions do not also have correspondingly high median household incomes, they tend to exhibit high home affordability ratio (indicating that housing is less affordable).

```python
fig = px.bar(
    df,
    x='key_row',
    y='median_household_income',
    title='Median Household Income by Region',
    labels={'key_row': 'Region', 'median_household_income': 'Median Household Income'},
)
fig.update_layout(xaxis_tickangle=-65)
fig.show()
```

The plot indicates that no region stands out with exceptionally high median household income, while Puerto Rico notably has one of the lowest median household incomes among all regions.

```python
fig = px.bar(
    df,
    x='key_row',
    y='median_sale_price',
    title='Median Sale Price by Region',
    labels={'key_row': 'Region', 'median_sale_price': 'Sale Price'},
)
fig.update_layout(xaxis_tickangle=-65)
fig.show()
```

The plot shows that regions such as California, Hawaii, Massachusetts, and Washington have notably high median sale prices, with Hawaii standing out at the top after California. As observed in previous charts, Hawaii also has a low population, which is interesting insight that may reflect the impact of high housing costs and geographic constraints on population density.

```python
fig = px.bar(
    df,
    x='key_row',
    y='house_affordability_ratio',
    title='House Affordability Ratio by Region (Lower is More Affordable)',
    labels={'key_row': 'Region', 'house_affordability_ratio': 'Affordability Ratio'}
)
fig.update_layout(xaxis_tickangle=-65)
fig.show()
```

The chart indicates that regions such as California, Hawaii, Montana, New York, and Washington have high house affordability ratios, suggesting that homes in these areas are significantly less affordable relative to household income.

```python
import plotly.express as px

fig = px.scatter(
    df,
    x='census_population',
    y='median_household_income',
    hover_name='key_row',
    size='median_sale_price',
    color='house_affordability_ratio',
    labels={
        'census_population': 'Population',
        'median_household_income': 'Household Income',
        'median_sale_price': 'Home Price',
        'house_affordability_ratio': 'Affordability Ratio'
    },
    title='Population vs. Household Income (Bubble size = Home Price)'
)
fig.show()
```

The chart reveals that there is a cluster of regions such as West Virginia, Mississippi, Arkansas, Oklahoma, Louisiana, Kentucky, Iowa, and Kansas tend to have both relatively low populations and lower household incomes, yet they achieve favorable housing affordability scores.
